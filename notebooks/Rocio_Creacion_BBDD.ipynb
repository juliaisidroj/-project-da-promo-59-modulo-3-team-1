```python
import mysql.connector
import pandas as pd

# 1. Configuración de la conexión
config = {
    'host': 'localhost',
    'user': 'root',
    'password': 'Mew33uga',
    'database': 'ABC_Corporation'
}

try:
    conn = mysql.connector.connect(**config)
    cursor = conn.cursor()

    # --- TABLA 1: EMPLEADOS ---
    cols_empleados = [
        'employee_number', 'age', 'gender', 'marital_status', 
        'education', 'education_field', 'distance_from_home', 'employee_count'
    ]
    # Filtramos el DataFrame para que coincida exactamente con la tabla
    data_empleados = df_abc[cols_empleados].values.tolist()
    
    query_empleados = """
    INSERT INTO empleados (employee_number, age, gender, marital_status, education, education_field, distance_from_home, employee_count)
    VALUES (%s, %s, %s, %s, %s, %s, %s, %s)
    """
    cursor.executemany(query_empleados, data_empleados)
    conn.commit()
    print(f"{cursor.rowcount} registros insertados en 'empleados'.")

    # --- TABLA 2: PUESTOS ---
    cols_puestos = [
        'employee_number', 'department', 'job_role', 'job_level', 'business_travel',
        'over_time', 'daily_rate', 'hourly_rate', 'monthly_income', 'monthly_rate',
        'stock_option_level', 'percent_salary_hike', 'num_companies_worked'
    ]
    data_puestos = df_abc[cols_puestos].values.tolist()
    
    query_puestos = """
    INSERT INTO puestos (employee_number, department, job_role, job_level, business_travel, over_time, daily_rate, hourly_rate, monthly_income, monthly_rate, stock_option_level, percent_salary_hike, num_companies_worked)
    VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s)
    """
    cursor.executemany(query_puestos, data_puestos)
    conn.commit()
    print(f"{cursor.rowcount} registros insertados en 'puestos'.")

    # --- TABLA 3: EVALUACION_Y_RETENCION ---
    cols_eval = [
        'employee_number', 'attrition', 'environment_satisfaction', 'job_involvement',
        'job_satisfaction', 'relationship_satisfaction', 'work_life_balance',
        'performance_rating', 'total_working_years', 'training_times_last_year',
        'years_at_company', 'years_in_current_role', 'years_since_last_promotion',
        'years_with_curr_manager'
    ]
    data_eval = df_abc[cols_eval].values.tolist()

    query_eval = """
    INSERT INTO evaluacion_y_retencion (employee_number, attrition, environment_satisfaction, job_involvement, job_satisfaction, relationship_satisfaction, work_life_balance, performance_rating, total_working_years, training_times_last_year, years_at_company, years_in_current_role, years_since_last_promotion, years_with_curr_manager)
    VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s)
    """
    cursor.executemany(query_eval, data_eval)
    conn.commit()
    print(f"{cursor.rowcount} registros insertados en 'evaluacion_y_retencion'.")

except mysql.connector.Error as err:
    print(f"Error: {err}")
finally:
    if conn.is_connected():
        cursor.close()
        conn.close()
```
